```python
import pandas as pd
import os

# 1. Cargar el archivo original
truck_events = pd.read_csv("../files/input/truck_event_text_partition.csv")

# 2. Definir specific_columns seleccionando solo algunas columnas
specific_columns = truck_events[['driverId', 'eventTime', 'eventType']]

# 3. Crear la carpeta de salida si no existe
if not os.path.exists("../files/output/"):
    os.makedirs("../files/output/")

# 4. Guardar el nuevo archivo CSV
specific_columns.to_csv(
    "../files/output/specific-columns.csv",
    sep=",",
    header=True,
    index=False,
)

print("¡Archivo procesado y guardado con éxito!")
```
